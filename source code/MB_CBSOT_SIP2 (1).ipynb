```python
import os
import numpy as np

drive_path = "/content/drive/MyDrive/cbsot ml/paper_embeddings.npy"
local_path = "paper_embeddings.npy"

if os.path.exists(drive_path):
    print("Loading saved embeddings from Google Drive")
    embeddings = np.load(drive_path)

elif os.path.exists(local_path):
    print("Loading uploaded embeddings file")
    embeddings = np.load(local_path)

else:
    print("Generating embeddings")
    embeddings = model.encode(
        df["paper_text"].tolist(),
        batch_size=32,
        show_progress_bar=True
    )
```

```python
drive_path = "/content/drive/MyDrive/cbsot ml/paper_faiss.index"
local_path = "paper_faiss.index"

if os.path.exists(drive_path):

    print("Loading existing FAISS index from Google Drive")
    index = faiss.read_index(drive_path)

elif os.path.exists(local_path):

    print("Loading uploaded FAISS index")
    index = faiss.read_index(local_path)

else:

    print("Creating new FAISS index")

    faiss.normalize_L2(embeddings)

    index = faiss.IndexFlatIP(384)
    index.add(embeddings)

    # Save locally
    faiss.write_index(index, local_path)

    print("FAISS index saved successfully!")
```

```python
def search_paper(query, k=5):
    query_embedding = model.encode([query])
    faiss.normalize_L2(query_embedding)
    D, I = index.search(query_embedding, k)
    for score, idx in zip(D[0], I[0]):
      print("Similarity Score", score)
      print("Title", df.iloc[idx]["title"])
      print("Abstract", df.iloc[idx]["abstract"][:500])
      print()
      # title = df.iloc[idx]["title"]
      # print(f"Title: {title}")
      # abstract = df.iloc[idx]["abstract"]
      # print(f"Abstract: {abstract}")
```

```python
def search_and_summarize(query, k=5):
    query_embedding = model.encode([query])
    faiss.normalize_L2(query_embedding)

    D, I = index.search(query_embedding, k)

    for score, idx in zip(D[0], I[0]):
        print("Similarity score", score)
        print("Title", df.iloc[idx]["title"])
        print("Abstract", df.iloc[idx]["abstract"][:500])
        print()

        summary = summarizer(
            df.iloc[idx]["abstract"],
            max_length=120,
            min_length=40,
            do_sample=False
        )

        print(summary)
        print(summary[0]["summary_text"])
        print()
```

```python
search_and_summarize("Deep learning in medical imaging", k = 5)
```
